# **icrawler**

**icrawler** provides a simple interface to automatically collect images from Bing using search keywords. By specifying a keyword and output directory, the library handles the complete pipeline of querying the search engine and downloading images, making it a practical tool for rapidly generating image datasets for machine learning and research projects.

```python
import keyword
import os
from icrawler.builtin import BingImageCrawler
```

### **2) Output path**

```python
# Directorio para guardar las imágenes descargadas
path = "images"
```

### **3) Download images**

```python
def download_images(path, keyword, max_num=50):
    """
    This function downloads images from Bing based on the provided keyword and saves them in the specified directory. 
    The number of images was set to 50 by default, but it can be adjusted by changing the max_num parameter.
    """
    if path not in os.listdir():
        os.mkdir(path)
    save_dir = os.path.join(path, keyword)
    bing_crawler = BingImageCrawler(storage={'root_dir': save_dir})
    bing_crawler.crawl(
        keyword=keyword,
        max_num=max_num
    )
    print(f"Descargadas {max_num} imágenes de {keyword} en {save_dir}")
```

```python
download_images(path, "induction kitchen", max_num=100)
download_images(path, "dirty induction kitchen", max_num=100)
```
